```python
import pygenelab as pgl
import scanpy as sc
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import decoupler as dc 
```

# Load DEGs

```python
# load DEGs

male_iix_degs_path = "/ocean/projects/cis240075p/asachan/datasets/TA_muscle/ERCC1_KO_mice/integrated_samples/analysis/male/degs/DEGs_KO_vs_WT_Fast_IIX.csv"
male_iib_degs_path = "/ocean/projects/cis240075p/asachan/datasets/TA_muscle/ERCC1_KO_mice/integrated_samples/analysis/male/degs/DEGs_KO_vs_WT_Fast_IIB.csv"
female_iix_degs_path = "/ocean/projects/cis240075p/asachan/datasets/TA_muscle/ERCC1_KO_mice/integrated_samples/analysis/female/degs/DEGs_KO_vs_WT_Fast_IIX.csv"
female_iib_degs_path = "/ocean/projects/cis240075p/asachan/datasets/TA_muscle/ERCC1_KO_mice/integrated_samples/analysis/female/degs/DEGs_KO_vs_WT_Fast_IIB.csv"

male_iix_degs = pd.read_csv(male_iix_degs_path)
male_iib_degs = pd.read_csv(male_iib_degs_path)
female_iix_degs = pd.read_csv(female_iix_degs_path)
female_iib_degs = pd.read_csv(female_iib_degs_path)
```

# Plot Volcano Plots

```python
male_iix_degs_fig, _ = pgl.plot_deg_volcano(
    deg_df=male_iix_degs,
    positive_label="KO Higher",
    negative_label="WT Higher",
    title="Male Fast IIX: WT vs KO"
)
```

```python
male_iib_degs_fig, _ = pgl.plot_deg_volcano(
    deg_df=male_iib_degs,
    positive_label="KO Higher",
    negative_label="WT Higher",
    title="Male Fast IIB: WT vs KO"
)
```

```python

female_iix_degs_fig, _ = pgl.plot_deg_volcano(
    deg_df=female_iix_degs,
    positive_label="KO Higher",
    negative_label="WT Higher",
    title="Female Fast IIX: WT vs KO"
)
```

```python
female_iib_degs_fig, _ = pgl.plot_deg_volcano(
    deg_df=female_iib_degs,
    positive_label="KO Higher",
    negative_label="WT Higher",
    title="Female Fast IIB: WT vs KO"
)
```

# Save As PNG

```python
pgl.save_fig_as_png(
    fig=male_iix_degs_fig,
    output_path="Output/M_FastIIX_Volcano.png"
)

pgl.save_fig_as_png(
    fig=male_iib_degs_fig,
    output_path="Output/M_FastIIB_Volcano.png"
)

pgl.save_fig_as_png(
    fig=female_iix_degs_fig,
    output_path="Output/F_FastIIX_Volcano.png"
)

pgl.save_fig_as_png(
    fig=female_iib_degs_fig,
    output_path="Output/F_FastIIB_Volcano.png"
)
```
